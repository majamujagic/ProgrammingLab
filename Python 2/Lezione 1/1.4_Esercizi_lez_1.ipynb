### **1. Trasformare cicli in list comprehension**

Fai tre esempi di cicli che producono una lista e poi riscrivi lo stesso codice usando la list comprehension

```python
%conda install numpy
import numpy as np
import random
```

```python
numeri = [2, 5, 8, 3, 9]

quadrati = []
for n in numeri:
    quadrati.append(n**2)
lista_quadrati = [n**2 for n in numeri]

pari = []
for n in numeri:
    pari.append(n%2==0)
lista_pari = [n for n in numeri if n%2==0]

piu_3 = []
for n in numeri:
    piu_3.append(n+3)
lista_piu_3 = [n+3 for n in numeri]
```
