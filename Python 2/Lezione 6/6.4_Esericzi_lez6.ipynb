# 5. Analisi di un dataset di video giochi
1. Quanti videogiochi sono stati pubblicati
2. Quali sono i generi più popolari? Mostralo con un bar plot
3. Plotta l'evoluzione del numero di giochi pubblicati nel tempo
4. Prova a riprodurre questo plot 




```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()
```

```python
totale_giochi = len(df)
giochi_unici = df['Name'].nunique()
totale_giochi, giochi_unici
```

```python
sns.barplot(data = df, x = 'Genre', y = 'Global_Sales', estimator=sum)
plt.title('Generi più popolari')
plt.xticks(rotation = 45)
plt.show
```

```python
giochi_per_anno = df.groupby('Year')['Name'].count()
plt.figure(figsize=(12, 6))
sns.lineplot(x=giochi_per_anno.index, y=giochi_per_anno.values, color='b', marker='o')
plt.title('Evoluzione del numero di videogiochi pubblicati nel tempo')
plt.xlabel('Anno')
plt.ylabel('Numero di giochi pubblicati')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
plt.figure(figsize=(12, 6))
top_platforms = df.groupby('Platform')['Global_Sales'].sum().sort_values(ascending=False).head(10)
sns.barplot(x=top_platforms.index, y=top_platforms.values, palette='viridis')
plt.title('Top 10 Piattaforme per Vendite Globali')
plt.xlabel('Piattaforma')
plt.ylabel('Vendite Globali (Milioni)')
plt.xticks(rotation=45)
plt.show()
```
